```python
i = 0
total_score = 0

while i >= 0:
    i += 1
    score = input("点数を入力してください:")
    print("点数を入力してください:"+score)

    if int(score) == -1:
        break

    total_score = int(score)+ total_score
    
rounded_average_score = round(total_score / (i-1), 1)
print(f"{i-1}人の平均点は{rounded_average_score}点です。")
```
