```python
price = input("税抜価格を入力してください:")
print("税抜価格を入力してください:"+price)
tax_price = round(int(price) * 1.1)
if tax_price >= 2000:
    print("送料は無料です")
    print("送料込みの価格は" + str(tax_price) + "円です。")
elif tax_price < 2000:
    print("送料として350円かかります")
    total_price = tax_price + 350
    print("送料込みの価格は" + str(total_price) + "円です。")
```
