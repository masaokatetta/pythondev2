```python
tango = {'a':0, 'b':0, 'c':0, 'd':0, 'e':0, 'f':0, 'g':0, 'h':0, 'i':0, 'j':0, 'k':0, 'l':0, 'm':0, 'n':0, 'o':0, 'p':0, 'q':0, 'r':0, 's':0, 't':0, 'u':0, 'v':0, 'w':0, 'x':0, 'y':0, 'z':0}
eitango = []
while True:
    word = input("英単語を入力してください：").lower()
    
    print(f"英単語を入力してください：{word}")

    eitango.append(word)

    if word == "":
        eitango.pop()
        print(f"入力された英単語：{eitango}")
        break

    for ch in word:
        if ch in tango:
            tango[ch] += 1

for ch, num in tango.items():
    if num >= 1:
        print(f"{ch}が{num}個ありました")
```
